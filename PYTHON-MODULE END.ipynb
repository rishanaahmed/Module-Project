```python
salary_expenditure = df.groupby(['Team', 'Position'])['Salary'].sum().reset_index()


max_salary_expenditure = salary_expenditure.loc[salary_expenditure['Salary'].idxmax()]

print(f"The highest salary expenditure is from Team {max_salary_expenditure['Team']} and Position {max_salary_expenditure['Position']} with a total expenditure of {max_salary_expenditure['Salary']}.")
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Team', y='Salary', hue='Position', data=salary_expenditure, palette='Set2')

plt.scatter(max_salary_expenditure['Team'], max_salary_expenditure['Salary'], color='red', s=100, label='Highest Expenditure', zorder=5)

plt.title('Salary Expenditure by Team and Position')
plt.xlabel('Team')
plt.ylabel('Salary Expenditure')
plt.legend()
plt.show()
```
